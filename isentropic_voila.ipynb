```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, BoundedFloatText, ToggleButtons
from isentropic_tools import (
    T0_T_relation,
    P0_P_relation,
    A_Astar_relation,
    solve_mach_from_area,
)
```

```python
# Function to plot isentropic flow properties
def isentropic_plot(k=1.4):
    M = np.linspace(0.1, 5, 500)

    T0_T = T0_T_relation(M, k)
    P0_P = P0_P_relation(T0_T, k)
    A_Astar = A_Astar_relation(M, k)

    fig, ax1 = plt.subplots(figsize=(5, 3))

    ax1.plot(M, T0_T, label='T₀/T', color='skyblue')
    ax1.plot(M, A_Astar, label='A/A*', color='lime')
    ax1.set_xlabel('Mach Number')
    ax1.set_ylabel('T₀/T and A/A*')
    ax1.grid(True)
    ax1.legend(loc='upper left')

    ax2 = ax1.twinx()
    ax2.plot(M, P0_P, label='P₀/P', color='orange')
    ax2.set_ylabel('P₀/P')
    ax2.legend(loc='upper center')

    plt.title(f'Isentropic Flow Properties vs Mach Number (k = {k})')
    plt.tight_layout()
    plt.show()

# Unified interface for both plot and Mach solver
def shared_k_interface(k=1.4, A_Astar_val=2.0, branch='subsonic'):
    isentropic_plot(k)

    M_result = solve_mach_from_area(A_Astar_val, k, branch)
    print(f"\nFor A/A* = {A_Astar_val}, k = {k}, and {branch} branch:")
    print(f"→ Mach number ≈ {M_result:.4f}")

# Interactive UI with one k slider
interact(
    shared_k_interface,
    k=FloatSlider(value=1.4, min=1.1, max=1.67, step=0.01, readout_format='.2f'),
    A_Astar_val=BoundedFloatText(value=2.0, min=1.01, max=5.0, step=0.0001),
    branch=ToggleButtons(options=['subsonic', 'supersonic'])
)
```
